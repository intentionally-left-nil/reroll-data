## Data setup

Connect to `data/main.db` (read-only; see `reroll_data.db2`) and pull the aggregate
counts needed to answer the questions above. `wheel` has one row per wheel filename
reroll's crawler has ever recorded; `reroll_data` is non-NULL exactly when reroll
converted that wheel successfully. `reroll_errors` has *at most one* row per wheel that
did **not** convert -- `category` is one of `scope` (deliberately out-of-scope wheel),
`invalid` (malformed wheel/metadata), `unconvertable` (in-scope, valid, but not
representable as a matchspec), `unavailable` (a data-completeness gap, e.g. PEP 658
metadata never fetched), `unexpected` (a bug), or `runtime` (an unstable-environment
error, recorded for accounting only -- see `reroll_data.db2`'s module docstring; it
never settles a wheel and currently has zero rows in this snapshot). A wheel with no
`reroll_errors` row and `reroll_data IS NULL` simply hasn't been attempted yet (backlog
on the `wheel_todo` partial index) -- unlike the legacy `v.db.repodata_conversion` table
this replaced, that state can actually occur now, since `main.db`/`pypi.db` are live and
continuously crawled/converted rather than a one-time migrated snapshot. "Supported" =
not `scope`; "convertible" = success (`reroll_data IS NOT NULL`). Every query below
reads only `main.db` -- `pypi.db` and the legacy `v.db` are never touched.

```python
import sqlite3
import re
import pandas as pd

MAIN_DB_PATH = "/Users/anil/code/reroll-data/data/main.db"


def query_main(sql, params=()):
    con = sqlite3.connect(f"file:{MAIN_DB_PATH}?mode=ro", uri=True)
    try:
        return pd.read_sql_query(sql, con, params=params)
    finally:
        con.close()


# Top-level category breakdown (cheap: a LEFT JOIN + GROUP BY over the whole
# wheel table, no row hydration). A wheel with no `reroll_errors` row has
# either succeeded (`reroll_data IS NOT NULL`) or not yet been attempted --
# same conflation the old `reroll_error IS NULL` bucket had against v.db,
# teased apart in the next cell.
category_counts = query_main(
    """
    SELECT
        COALESCE(e.category, 'success') AS category,
        COUNT(*) AS n
    FROM wheel w
    LEFT JOIN reroll_errors e ON e.wheel_id = w.id
    GROUP BY category
    """
)
total = int(category_counts["n"].sum())
category_counts["pct_of_all"] = (category_counts["n"] / total * 100).round(2)
print(f"total rows (all wheels main.db.wheel has a row for): {total:,}")
category_counts.sort_values("n", ascending=False)
```

```python
# The "success" bucket above conflates two things, same as the legacy
# v.db.repodata_conversion analysis: rows reroll has actually converted
# (`reroll_data IS NOT NULL`) vs. rows not yet attempted at all
# (`reroll_data IS NULL`, no `reroll_errors` row -- backlog on the
# `wheel_todo` index). `yanked` is split out separately (not folded into
# `not_yet_attempted`): `reroll_data.reroll_convert`'s own worklist
# (`db2.OUTSTANDING_WHEEL` / the `wheel_todo` partial index) deliberately
# excludes `yanked = 1` rows -- "nothing consumes a yanked wheel's
# `reroll_data` under normal dependency resolution, so it is not worklist
# work" (see `reroll_data.db2`'s `wheel_todo` comment) -- so a yanked wheel
# sits at `reroll_data IS NULL` forever by design, and lumping it into
# `not_yet_attempted` overstated the real backlog by ~118.6k rows (all of
# `not_yet_attempted` turned out to be yanked wheels in this snapshot; see
# the notebook's own diagnosis of that miscount). Also confirm `category`
# never holds anything outside the fixed CHECK enum on `reroll_errors`
# (scope/invalid/unconvertable/unavailable/unexpected/runtime) -- unlike the
# old free-text `reroll_error` column, there is no "other/unrecognized"
# bucket possible here by construction, but check anyway since this is
# live, evolving data.
status_counts = query_main(
    """
    SELECT
        CASE
            WHEN e.category IS NOT NULL THEN e.category
            WHEN w.reroll_data IS NOT NULL THEN 'success'
            WHEN w.yanked = 1 THEN 'yanked'
            ELSE 'not_yet_attempted'
        END AS status,
        COUNT(*) AS n
    FROM wheel w
    LEFT JOIN reroll_errors e ON e.wheel_id = w.id
    GROUP BY status
    """
)
status_counts["pct_of_all"] = (status_counts["n"] / total * 100).round(2)
print(status_counts.sort_values("n", ascending=False).to_string(index=False))

_KNOWN_STATUSES = {
    "scope",
    "invalid",
    "unconvertable",
    "unavailable",
    "unexpected",
    "runtime",
    "success",
    "not_yet_attempted",
    "yanked",
}
_unknown = set(status_counts["status"]) - _KNOWN_STATUSES
assert not _unknown, f"unexpected reroll_errors.category value(s): {_unknown}"
```

```python
# Definitions (matching reroll's own taxonomy in reroll.errors /
# docs/errors_and_logging.md, now backed directly by
# main.db.reroll_errors.category instead of a parsed text prefix):
#   unsupported  = scope + invalid   (RerollScopeError / RerollInvalidWheelError)
#   attempted    = total - not_yet_attempted - yanked  (wheels reroll has actually classified)
#   supported    = attempted - unsupported = success + unconvertable + unexpected + unavailable + runtime
#   Q1: % of supported wheels reroll CAN convert  = success / supported
#   Q2: % of all wheels reroll supports            = supported / attempted
#   O2: % of all wheels reroll does NOT support    = unsupported / attempted  (100 - Q2)
#   Q4: count of supported-but-unconvertable wheels = unconvertable
#   Q5 buckets: % of ALL SUPPORTED wheels (not just % of the unconvertable slice)
#
# `not_yet_attempted` and `yanked` are both excluded from every denominator
# below -- `not_yet_attempted` is genuine backlog on `wheel_todo` (Q2/O2
# therefore treat the attempted wheels as representative of the full PyPI
# population, a fair assumption while the backlog stays a small, unbiased
# slice -- see the not_yet_attempted % printed below); `yanked` is not
# backlog at all -- `reroll_convert`'s own worklist (`wheel_todo`)
# deliberately never attempts a yanked wheel (see the status_counts cell
# above), so it would never move out of this bucket no matter how many
# `convert` runs happen, and including it in `attempted` would permanently
# understate Q1/Q2. `runtime` (a RerollRuntimeError, recorded for accounting
# only, never settles a wheel -- see `reroll_data.db2`'s module docstring)
# is currently zero rows in this snapshot but is folded into `supported`
# rather than dropped, matching `stats_main`'s own accounting.
def _n(status):
    m = status_counts.loc[status_counts.status == status, "n"]
    return int(m.iloc[0]) if len(m) else 0


n_not_yet_attempted = _n("not_yet_attempted")
n_yanked = _n("yanked")
n_success = _n("success")
n_scope = _n("scope")
n_invalid = _n("invalid")
n_unconvertable = _n("unconvertable")
n_unexpected = _n("unexpected")
n_unavailable = _n("unavailable")
n_runtime = _n("runtime")

attempted = total - n_not_yet_attempted - n_yanked
n_unsupported = n_scope + n_invalid
n_supported = attempted - n_unsupported
assert (
    n_supported
    == n_success + n_unconvertable + n_unexpected + n_unavailable + n_runtime
)

Q1 = n_success / n_supported * 100
Q2 = n_supported / attempted * 100
O2 = n_unsupported / attempted * 100
Q4 = n_unconvertable

print(f"total wheels in main.db.wheel:  {total:>12,}")
print(
    f"  yanked:                        {n_yanked:>10,}  "
    f"({n_yanked / total * 100:.2f}% -- excluded from all % below; "
    "reroll_convert never attempts these by design)"
)
print(
    f"  not yet attempted:            {n_not_yet_attempted:>10,}  "
    f"({n_not_yet_attempted / total * 100:.2f}% -- excluded from all % below)"
)
print(f"attempted:                      {attempted:>12,}")
print(f"  unsupported (scope+invalid):  {n_unsupported:>10,}  ({O2:.2f}% = O2)")
print(f"  supported:                    {n_supported:>10,}  ({Q2:.2f}% = Q2)")
print(
    f"    success:                    {n_success:>10,}  ({Q1:.4f}% of supported = Q1)"
)
print(f"    unconvertable:              {n_unconvertable:>10,}  (Q4)")
print(f"    unexpected (bugs):          {n_unexpected:>10,}")
print(f"    unavailable (data gap):     {n_unavailable:>10,}")
print(f"    runtime (accounting only):  {n_runtime:>10,}")
```

### Q3: what makes up the {O2}% of "unsupported" wheels

`scope` (deliberate policy exclusions) + `invalid` (malformed wheel/metadata) errors,
broken into exception-type sub-buckets. `main.db.reroll_errors.category`/`.sub_category`
are already split columns (`sub_category` is the raising exception's class name), so
this is a plain `GROUP BY category, sub_category` over the ~1.7M-row `scope`/`invalid`
slice of `reroll_errors` -- no more regex/UDF parsing of a combined `"{category}:
{ExcType}: {message}"` text column, unlike the legacy
`v.db.repodata_conversion.reroll_error` this replaced.

```python
scope_invalid_types = query_main(
    """
    SELECT category, sub_category AS exc_type, COUNT(*) AS n
    FROM reroll_errors
    WHERE category IN ('scope', 'invalid')
    GROUP BY category, exc_type
    ORDER BY n DESC
    """
)
scope_invalid_types["pct_of_attempted"] = (
    scope_invalid_types["n"] / attempted * 100
).round(3)
print(f"sum of n: {scope_invalid_types['n'].sum():,} (should equal {n_unsupported:,})")
scope_invalid_types
```

```python
# One representative message per exc_type, to write accurate human
# descriptions for the Q3 table. sub_category/description are already real
# columns, so this is a plain window-function pass -- no UDF needed.
samples = query_main(
    """
    WITH ranked AS (
        SELECT
            sub_category AS exc_type,
            description AS reroll_error,
            ROW_NUMBER() OVER (PARTITION BY sub_category ORDER BY wheel_id) AS rn
        FROM reroll_errors
        WHERE category IN ('scope', 'invalid')
    )
    SELECT exc_type, reroll_error FROM ranked WHERE rn = 1
    """
)
pd.set_option("display.max_colwidth", 300)
samples
```

### Q4 / Q5: the unconvertable bucket

Same targeted `category = 'unconvertable'` fetch (small population, cheap) as before,
now against `main.db.reroll_errors` -- rerun live rather than reusing copied-verbatim
counts, which go stale the moment the bucket's size or composition changes between
runs. It has, in fact, changed shape since the `v.db`-era classification: `main.db`'s
new curated-mapping gate (`reroll_data.reroll_convert`'s `MissingPypiCondaStaticMapping`,
a `RerollUnconvertableError` subclass for a dependency name reroll's live mapper chain
resolved but that no human has yet curated into `main.db.pypi_conda_names`) is now the
single largest `unconvertable` sub-bucket, and did not exist at all in the old
classification.

```python
# Load the full "unconvertable" bucket live (small population -- a targeted
# fetch, not a scan of the 12.3M-row wheel table) and re-run the same
# classification regexes `reroll_v0.2.0_unconvertable_analysis.ipynb`
# established, now against the already-split sub_category/description
# columns instead of a combined "unconvertable: <ExcType>: <message>" text
# column -- no exc_type-extraction regex needed anymore.
unconvertable_df = query_main(
    "SELECT wheel_id, sub_category AS exc_type, description AS reroll_error "
    "FROM reroll_errors WHERE category = 'unconvertable'"
)

# UnconvertableRequirementError sub-buckets
req_errs = unconvertable_df[
    unconvertable_df["exc_type"] == "UnconvertableRequirementError"
].copy()


def classify_req_error(msg):
    m = re.search(r"cannot convert '[^']*':\s*(.+)$", msg)
    if m:
        return m.group(1)
    m = re.search(r"is not a legal conda extra name \(CEP-29\):\s*(.+)$", msg)
    if m:
        return f"illegal conda extra name: {m.group(1)}"
    return "OTHER/unrecognized"


req_errs["reason_template"] = (
    req_errs["reroll_error"]
    .map(classify_req_error)
    .str.replace(r"'[^']*'", "<LIT>", regex=True)
    .str.replace(r"\b\d+(?:\.\d+)*\b", "<NUM>", regex=True)
)
req_sub = req_errs["reason_template"].value_counts()

# UnconvertableMarkerError sub-buckets
marker_errs = unconvertable_df[
    unconvertable_df["exc_type"] == "UnconvertableMarkerError"
].copy()
unpermitted = marker_errs["reroll_error"].str.extract(
    r"unpermitted key\(s\) \['([^']*)'"
)[0]


def bucket_marker(msg, key):
    if pd.notna(key):
        return "unpermitted key"
    if "markers are not supported" in msg:
        return "'in'/'not in' marker unsupported"
    if "not a valid version" in msg:
        return "python_version literal not a valid version"
    if "is not supported for" in msg:
        return "unsupported comparator (~=) for python_version(_full)"
    if "not a plain major.minor" in msg:
        return "python_version literal not major.minor precision"
    return "OTHER/unrecognized"


marker_errs["bucket"] = [
    bucket_marker(m, k) for m, k in zip(marker_errs["reroll_error"], unpermitted)
]
marker_sub = marker_errs["bucket"].value_counts()

# UnresolvedCondaNameError sub-buckets
unres = unconvertable_df[
    unconvertable_df["exc_type"] == "UnresolvedCondaNameError"
].copy()
unres["dep_name"] = unres["reroll_error"].str.extract(
    r"no mapper resolved a conda name for '([^']*)'"
)[0]
unres_sub = unres["dep_name"].value_counts()

# InvalidCondaNameError sub-buckets
invalid_errs = unconvertable_df[unconvertable_df["exc_type"] == "InvalidCondaNameError"]
n_long_names = invalid_errs["reroll_error"].str.contains("exceeds 64 characters").sum()
n_homoglyph = len(invalid_errs) - n_long_names

n_pyver_eq = (
    unconvertable_df["exc_type"] == "UnconvertablePythonVersionEqualityError"
).sum()

# MissingPypiCondaStaticMapping -- new in the db2 schema, does not appear in
# any v.db-era snapshot: reroll_data.reroll_convert's curated-mapping gate
# rejects a wheel whose live mapper chain resolved a dependency name reroll
# has not yet had a human curate into main.db.pypi_conda_names (see that
# module's docstring, "The new mapper" section). No further
# sub-classification needed -- description is already specific (one
# unresolved pypi name per row).
n_missing_static = int(
    (unconvertable_df["exc_type"] == "MissingPypiCondaStaticMapping").sum()
)

q5_rows = [
    (
        "MissingPypiCondaStaticMapping",
        "dependency name reroll's live mapper chain resolved, but no human has curated it into main.db.pypi_conda_names yet",
        n_missing_static,
    ),
    (
        "UnconvertableRequirementError",
        "PEP 440 local version label (torch/ipex-llm/vllm-cpu +cpu/+xpu hardware variants)",
        int(
            req_sub.get(
                "cannot convert specifier <LIT>: it has a local version label", 0
            )
        ),
    ),
    (
        "InvalidCondaNameError",
        "name >64 chars (SEO-spam PyPI project names)",
        int(n_long_names),
    ),
    (
        "UnconvertableRequirementError",
        "illegal conda extra name (CEP-29)",
        int(
            req_sub.get(
                "illegal conda extra name: must be <NUM>-<NUM> characters of [a-z0-9_.+-]",
                0,
            )
        ),
    ),
    (
        "UnconvertableMarkerError",
        "unsupported comparator (~=) for python_version(_full)",
        int(marker_sub.get("unsupported comparator (~=) for python_version(_full)", 0)),
    ),
    (
        "UnconvertableMarkerError",
        "python_version literal not a valid version (e.g. '2.7.*', '3.x', 'dev')",
        int(marker_sub.get("python_version literal not a valid version", 0)),
    ),
    (
        "UnconvertableMarkerError",
        "unpermitted key (platform_release/_version/_machine/sys_platform/platform_python_implementation)",
        int(marker_sub.get("unpermitted key", 0)),
    ),
    (
        "UnconvertableRequirementError",
        "amulet-compiler-version numeric-overflow version",
        int(req_sub.get("OTHER/unrecognized", 0)),
    ),
    (
        "UnconvertableMarkerError",
        "python_version literal not major.minor precision",
        int(marker_sub.get("python_version literal not major.minor precision", 0)),
    ),
    (
        "UnconvertableMarkerError",
        "'in'/'not in' marker unsupported",
        int(marker_sub.get("'in'/'not in' marker unsupported", 0)),
    ),
    (
        "UnconvertablePythonVersionEqualityError",
        "python_version ==/!= a non-major.minor literal",
        int(n_pyver_eq),
    ),
    (
        "UnconvertableRequirementError",
        "direct URL reference",
        int(req_sub.get("it has a direct URL reference", 0)),
    ),
    (
        "UnresolvedCondaNameError",
        f"{len(unres_sub)} unresolved dep names ({', '.join(unres_sub.index)})",
        int(unres_sub.sum()),
    ),
    (
        "InvalidCondaNameError",
        "homoglyph/Cyrillic lookalike name (e.g. `tаamo`)",
        int(n_homoglyph),
    ),
]
q5 = pd.DataFrame(q5_rows, columns=["exc_type", "description", "n"])
q5 = q5[q5["n"] > 0].sort_values("n", ascending=False).reset_index(drop=True)

assert q5["n"].sum() == n_unconvertable == Q4, (q5["n"].sum(), n_unconvertable, Q4)
q5["pct_of_supported"] = (q5["n"] / n_supported * 100).round(3)
print(f"Q4 = {Q4:,} rows")
print(
    f"sum of pct_of_supported = {q5['pct_of_supported'].sum():.2f}% "
    f"(should equal Q4/n_supported = {Q4 / n_supported * 100:.2f}%)"
)
q5
```
