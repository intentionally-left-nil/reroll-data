```python
import sqlite3, time, json

DB_PATH = "data/v.db"

con = sqlite3.connect(DB_PATH)
con.execute("PRAGMA query_only = ON;")
con.execute("PRAGMA cache_size = -1000000;")  # ~1GB page cache
con.execute("PRAGMA temp_store = MEMORY;")
```

```python
q = """
SELECT COUNT(*)
FROM wheel
WHERE filename LIKE '%-none-any.whl';
"""
total_pure_python = con.execute(q).fetchone()[0]
total_pure_python
```

```python
q = """
SELECT COUNT(*)
FROM wheel w
JOIN metadata_blob mb ON mb.sha256 = w.metadata_sha256
WHERE w.filename LIKE '%-none-any.whl'
  AND (
    mb.parsed_json LIKE '%;%os_name%'
    OR mb.parsed_json LIKE '%;%platform_system%'
    OR mb.parsed_json LIKE '%;%sys_platform%'
    OR mb.parsed_json LIKE '%;%platform_machine%'
  );
"""
t0 = time.time()
wheel_count = con.execute(q).fetchone()[0]
print(wheel_count, time.time() - t0)
```
